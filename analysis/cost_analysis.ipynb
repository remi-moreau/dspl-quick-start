# Cost analysis

We take `barcode03_genscript` and JPEGDNA $\Delta G$ (Bird).

## Writing costs

```python
# WRITING COST
WRITE_COST_EURO_PER_NT = 0.02

BIRD_SIZE_IN_BYTES = 852200

JPEGDNA_BIRD_N_REF = 1983

MOTIF_BIRD_N_REF = 1698

BIRD_FORMATTED_OLIGO_SIZE = 72*2+50 # 202


def write_cost_per_byte(image_size_in_bytes, n_ref, formatted_oligo_size, write_cost_euro_per_nt):
    n_nt = n_ref * formatted_oligo_size
    write_cost = write_cost_euro_per_nt * n_nt
    return write_cost / image_size_in_bytes


print("Write cost per MB with JPEGDNA delta G: ", write_cost_per_byte(BIRD_SIZE_IN_BYTES, JPEGDNA_BIRD_N_REF, BIRD_FORMATTED_OLIGO_SIZE, WRITE_COST_EURO_PER_NT) * 1e6, "€")
print("Write cost per MB with MOTIF PAIRCODE: ", write_cost_per_byte(BIRD_SIZE_IN_BYTES, MOTIF_BIRD_N_REF, BIRD_FORMATTED_OLIGO_SIZE, WRITE_COST_EURO_PER_NT) * 1e6, "€")
```

## Reading costs

```python
# READING COSTS

READ_COST_EURO_PER_READ = 0.01/1000  # 0.00001

JPEGDNA_BIRD_AVG_DECODING_COVERAGE = 7.73

MOTIF_BIRD_AVG_DECODING_COVERAGE = 320

def read_cost_per_byte(image_size_in_bytes, n_ref, avg_decoding_coverage, read_cost_euro_per_read):
    avg_n_reads_to_decode = n_ref * avg_decoding_coverage
    read_cost = read_cost_euro_per_read * avg_n_reads_to_decode
    return read_cost / image_size_in_bytes

print("Read cost per MB with JPEGDNA delta G: ", read_cost_per_byte(BIRD_SIZE_IN_BYTES, JPEGDNA_BIRD_N_REF, JPEGDNA_BIRD_AVG_DECODING_COVERAGE, READ_COST_EURO_PER_READ) * 1e6, "€")
print("Read cost per MB with MOTIF PAIRCODE: ", read_cost_per_byte(BIRD_SIZE_IN_BYTES, MOTIF_BIRD_N_REF, MOTIF_BIRD_AVG_DECODING_COVERAGE, READ_COST_EURO_PER_READ) * 1e6, "€")
```

Total

```python
jpegdna_delta_g_total_cost_per_mb = (write_cost_per_byte(BIRD_SIZE_IN_BYTES, JPEGDNA_BIRD_N_REF, BIRD_FORMATTED_OLIGO_SIZE, WRITE_COST_EURO_PER_NT) + read_cost_per_byte(BIRD_SIZE_IN_BYTES, JPEGDNA_BIRD_N_REF, JPEGDNA_BIRD_AVG_DECODING_COVERAGE, READ_COST_EURO_PER_READ)) * 1e6
motif_paircode_total_cost_per_mb = (write_cost_per_byte(BIRD_SIZE_IN_BYTES, MOTIF_BIRD_N_REF, BIRD_FORMATTED_OLIGO_SIZE, WRITE_COST_EURO_PER_NT) + read_cost_per_byte(BIRD_SIZE_IN_BYTES, MOTIF_BIRD_N_REF, MOTIF_BIRD_AVG_DECODING_COVERAGE, READ_COST_EURO_PER_READ)) * 1e6

print("Total cost with JPEGDNA delta G: ", jpegdna_delta_g_total_cost_per_mb, "€")
print("Total cost with MOTIF PAIRCODE: ", motif_paircode_total_cost_per_mb, "€")
print("Motif PAIRCODE is better than JPEGDNA delta G by (%):", (jpegdna_delta_g_total_cost_per_mb - motif_paircode_total_cost_per_mb) / jpegdna_delta_g_total_cost_per_mb * 100, "%")
```
